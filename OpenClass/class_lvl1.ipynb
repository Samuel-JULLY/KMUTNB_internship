## Application of machine learning

Let’s imagine you are the data manager for an online content platform whose business model relies on the number of subscriptions. You are asked two questions:

- What is the profile of users who subscribe?

- How can we predict whether a new user will subscribe?

In both cases, the goal is to better understand the act of subscribing. However, these questions require two different approaches:

- Statistical modeling

- Predictive modeling

### Stastical modelisation

The first question (“What is the profile of users who subscribe?”) is about user profiles. With this question, the goal is to understand the act of subscribing based on user characteristics.

This involves an analytical and interpretative approach to exploring the relationships between variables. Statistical modeling relies on hypothesis testing and mathematical models to evaluate the conclusions of the analysis.

### Predictive modelisation

The second question (“How can we predict whether a new user will subscribe?”) is about prediction. With this question, the goal is solely to predict the act of subscribing as effectively as possible. This is the Machine Learning approach.

We expect two things from a predictive model based on a Machine Learning approach:

High-quality predictions, meaning a good score measured using a previously chosen metric.

The ability to extrapolate, that is, to generalize its predictions from the training data. This refers to the model’s robustness when faced with new samples.

In this course, we will focus on what is known as classical Machine Learning. Classical ML works with tabular datasets, meaning data that can be represented in table form and that are of a reasonable size (less than 1 gigabit). Such a dataset can come in the form of a simple CSV, Excel, or JSON file, or be directly extracted from a database (SQL, NoSQL).

Heavier data types—such as audio, images, or video—as well as generative AI, are left to deep learning and neural networks.

At the core of Machine Learning lies the dataset. Learning is driven by data.
No data, no Machine Learning!

### Understand what is machine learning

#### Working on the data

It all starts with transforming raw data to make it compatible with the chosen ML model: cleaning, normalization, digitization, etc. This is referred to as data cleaning.

Examples include:

- Missing data (e.g., half of the height values are missing),

- Incorrect data (e.g., someone is 200 years old or a weight is negative),

- Need for normalization (e.g., house prices in hundreds of thousands vs. area in the hundreds),

- Data format and accessibility issues (e.g., 200 Excel files to combine, files too large, incompatible formats, etc.).

#### Cross-validation

Since we want to build a model that performs well on data it hasn’t seen during training, we need to avoid evaluating the model on the training samples.

So, we split the dataset into two parts:

- One part of the samples is used to train the model. By convention, this is called the training set or simply train.

- The other part is set aside to evaluate the model’s performance on unseen data. This is called the test set.

A typical train/test split ratio is 80/20%.

In fact, we often repeat this splitting process multiple times and randomly to ensure the model performs well regardless of how the data is divided. This technique is called cross-validation (which we’ll cover in more detail later).



Over time, as we experiment, we risk manually optimizing the model based on the test dataset. To avoid this, we split the dataset not into 2 but into 3 parts, ensuring that a portion of the data remains completely unseen by the model.

So, before performing cross-validation, we set aside a validation set. This validation set is only used to evaluate the model’s performance once it has been fully trained and optimized.

However, in some application domains—such as stock prediction—we need to be absolutely certain that this separation is strict. In such cases, we have only one shot at using the validation set. Once that "shot" is taken, we no longer touch the model, for fear of contaminating or biasing it.

That said, this is an extreme case of separating model training from evaluation.

#### Optimization

At the same time, we aim to improve the model’s performance by tuning its parameters and observing its score on each train/test split of the dataset.

Each type of model has its own set of parameters, which vary depending on the libraries used.

It’s through hands-on work with these parameters that one develops real data science expertise.

### Difference from supervised and non-supervised approch

In Machine Learning, it’s important to distinguish between:

- The unsupervised approach

- The supervised approach

Imagine you have a large collection of photos of cats and dogs. You want to automatically classify them using a Machine Learning model.

#### The Unsupervised Approach
In the unsupervised approach, you don't have any labeling system to distinguish between photos of cats and dogs.

However, the model will try to automatically group the photos based on their intrinsic characteristics. It clusters them into groups with no prior knowledge of the species.

This is known as clustering.

#### The Supervised Approach
In contrast, the supervised approach requires that each image be labeled as either a “cat” or a “dog.” This label becomes the target variable the model will try to predict.

This labeling is done by a human and can involve a significant cost, especially when dealing with thousands—or even millions—of samples.

In supervised learning, depending on the nature of the target variable, we distinguish between:

- Classification, when predicting categories:

  - Binary (yes/no),

  - Ordinal (small, medium, large),

  - or Nominal (cat – dog – sheep, or blue – pink – green – white...)

- and Regression, when predicting a continuous variable: price, age, salary, volume, temperature, etc.

For example, let's take a dataset containing the age, height, and weight of around a hundred middle school students. If we want to predict the children's weight based on their height and age, the target variable will be the weight, and the predictive variables will be age and height.

We start from a dataset that is, in principle, “usable.” This is a catch-all concept implying that the variables are related to one another. The dataset is not just a totally random collection of data gathered without any structure or logic.

Returning to our example with age, height, and weight: it is reasonable to assume that a child's height and age are related to their weight.

So here is our formula:

weight = function(age, height) + noise

The "noise" represents the part of the information that is not captured by the model.

Here is a dataset with 237 samples including age, sex, height, and weight of children aged 11.5 to 20 years.

Height in cm ranges from 128.27 cm to 182.88 cm.

## Predictive models

Let’s take a closer look at what we mean by a predictive model. There are many types of models, each with its own background, characteristics, and performance.

Since we can't test them all one by one, you need to be able to quickly choose the best tools to understand your dataset.

As a general rule, you’ll choose the model based on the context and the project requirements:

- the task to perform: regression, classification, or clustering;

- the nature of the data: numerical, categorical, text, images, video, audio, time series;

- the volume of data;

- the statistical properties of the data: linearity, stationarity, range, completeness, consistency, etc.

To save time, we now distinguish between the following main model classes:

- Regression models

- Tree-based models

- Neural networks and Deep Learning

### Differents type of models

#### Regression model

Suitable for datasets of limited size where the relationships between variables are linear, at least to some extent. This is also referred to as the Generalized Linear Model (GLM), a family of models that includes linear regression in all its forms, as well as variants for classification tasks.

We refer to them as regression models, but that doesn't mean they are limited to regression problems! As we will see in Part 2, regression models can be used to solve both regression problems (predicting continuous values) and classification problems (predicting discrete categories).

#### Trees based model

Decision trees, random forests, as well as XGBoost and its many variants, are well-suited for larger tabular datasets. They are highly robust to missing data and outliers. These models are by far the top performers in relevant Kaggle competitions.

#### Neuronal network and Deep learning

Suited for large-scale datasets of more complex nature, such as images or video. These models are much more versatile in the types of data they can handle (images, video) and provide significant performance gains when applied to large volumes of data.

Other types of models, while certainly relevant in specific contexts, do not offer the same versatility and robustness as these three types of models.

Improvement #1: Add a quadratic term

Quadratic simply means "squared."

If we closely look at the scatter plot of sales versus TV, we notice that the points tend to follow a curve rather than a straight line. This is especially true for the lower-left part of the graph.


We can deduce that the relationship between sales and TV is not simply linear, that is:

$
\text{sales} = a \times \text{TV} + b
$

But it also depends on a TV² term, like this:

$
\text{sales} = a \times \text{TV} + b \times \text{TV}^2 + c
$

This is a second-degree polynomial in the variable TV.

When we add a power of one of the variables in the regression, this is called polynomial regression.

Polynomial regression consists of adding powers of certain predictors in the regression. It is a simple way to capture nonlinear relationships between variables.

Before training this new model, we notice that the range of TV² will be much larger than that of the other variables. So, we will normalize the variables to make their ranges compatible.

Scikit-learn offers different methods to normalize or standardize variables.

Let's create the new variable TV² = TV^2:


```python
# Create the quadratic variable tv^2
df['tv2'] = df.television**2

variables = ['television','radio','newspaper','tv2']
```

Let's normalize the range using MinMaxScaler, which forces the variables to be between 0 and 1.


```python
from sklearn.preprocessing import MinMaxScaler
# Instanciate the Scaler
scaler = MinMaxScaler()

# apply
data_array = scaler.fit_transform(df[variables])

# conversion of array to dataframe
df_scaled = pd.DataFrame(data_array,  columns = variables)

# Check that the min and max values of the variables are all 0 and 1
df_scaled.describe().loc[['min','max']]
```

At this stage, `df` contains only values between 0 and 1.


```python
# splitting data in train and test
X = df_scaled[variables]
y = df.sales

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
# training the model
reg = LinearRegression()
reg.fit(X_train, y_train)
```

```python
# Prediction on the test set
y_pred_test = reg.predict(X_test)

# Scores
print(f"-- sales regression ~ radio + newspaper + tv + tv^2")
print(f"\tRMSE: {mean_squared_error(y_test, y_pred_test)}")
print(f"\tMAPE: {mean_absolute_percentage_error(y_test, y_pred_test)}")
```

```python
# let's create the variable tv * radio

df['tv_radio'] = df.television * df.radio
```

```python
regressions = {
    'simple: y ~ television + radio + newspaper'  :     ['television','radio','newspaper'],
    'quadratique: y ~ television + radio + newspaper + tv2': ['television','radio','newspaper', 'tv2'],
    'cross term: y ~ television + radio + newspaper + tv*radio':['television','radio','newspaper', 'tv_radio']
}
```

```python
# the target is the same
y = df.sales


# loop on the regressions
for title, variables in regressions.items():
    # it is necessary to limit the amplitude on all predictor variables
    scaler = MinMaxScaler()
    data_array = scaler.fit_transform(df[variables])
    # df_scaled = pd.DataFrame(data_array, columns = variables)
    
    # X = df_scaled[variables]
    # Train test split
    X_train, X_test, y_train, y_test = train_test_split(data_array, y, test_size=0.20, random_state=42)

    # train the model
    reg.fit(X_train, y_train)
    # prediction on test set
    y_pred_test = reg.predict(X_test)

    # Scores
    print(f"\n-- Regression {title}")
    print(f"\tRMSE: {mean_squared_error(y_test, y_pred_test)}")
    print(f"\tMAPE: {mean_absolute_percentage_error(y_test, y_pred_test)}")
```

We observe a clear improvement compared to the first model.

It is possible to go even further by adding the interaction term `tv * radio`, which reflects the combined effect of an advertising campaign on both radio and television.


## Classify the data using logistic regression

### Discover the principle of logistic regression (or classification)

There are several types of classification depending on the values taken by the target variable:

- Binary classification, where the target variable takes 2 exclusive values: yes/no, 0/1, true/false, works/doesn’t work, etc.

- Multiclass classification, where the target variable can take more than 2 values. Within this case, we distinguish:

    - Ordinal classification, where the values are ordered: 1, 2, 3; bad, average, good;

    - Nominal classification, where the categories are not ordered: colors, nationalities, animals...

Finally, we speak of multilabel classification when a sample can belong to multiple categories at the same time. For example, a movie’s genres or an animal’s taxonomy: a tiger is at once a mammal, a feline, and a carnivore.

We will mainly focus on the simplest case: binary classification.

### Understand the relationship between classification and regression

The general principle of modeling is similar to that of regression; what mainly changes are the model scoring metrics and their interpretation.

But wait, why do we talk about logistic regression for classification?

Actually, we adapt linear regression to the classification case by interpreting the linear regression prediction as one of the categories of the target variable.

Take the case of a linear regression with one predictor:

$ y=ax+b $

Here, $y$ is continuous and can potentially take any real value.

Imagine a function $f$ that projects $y$ into the interval $[0,1]$ :

$z=f(y)=f(ax+b)$

We can interpret the variable $𝑧$ as the probability that the prediction belongs to one of the categories, considering, for example, a classification threshold of 0.5.

- If $f(y) < 0.5$, then class 0;

- If $f(y) ≥ 0.5$, then class 1;

A number between 0 and 1 can be interpreted as a probability because it measures the likelihood of an event occurring. 0 means impossible, 1 means certain, and values between 0 and 1 indicate relative chances of the event.

So, we can apply linear regression, project the result into the interval $[0,1]$, and interpret this value as the probability of belonging to one of the desired categories.

More precisely, a **logistic regression** in the case of **binary classification** consists of:

1. **Model a linear regression**:  
   $   y = a \cdot x + b   $

2. **Project the prediction \( y \)** into the interval \([0, 1]\) using the **logistic function**:  
   $
   \text{logistic}(y) = \frac{1}{1 + e^{-y}}
   $

3. **Interpret** \( \text{logistic}(y) \) as the **probability** that the outcome belongs to one class or the other, using a **threshold** \( t = 0.5 \):

   - If $ \text{logistic}(y) < 0.5 $ → class **0**
   - If $ \text{logistic}(y) \geq 0.5 $ → class **1**

This approach transforms a linear regression into a probabilistic classification model.

### Evaluate the performance of a classification model

The simplest way to measure the performance of a classification model is to look at the number of samples that were correctly classified. This is accuracy.

$$
\text{exactitude} = \frac{\text{correctly classified samples}}{\text{total number of samples}}
$$


```python
y_pred = clf.predict(X)

from sklearn.metrics import accuracy_score
accuracy_score(y, y_pred)
```

95 % ! That's a good score for a simple logistic regression

Is this model biased or objective? Does it tend to classify 0s as 1s or 1s as 0s?

To see this, we use the confusion matrix — a 2x2 table with the true values as columns and the predicted values as rows:

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y, y_pred)
```

In general, in the binary case, the confusion matrix is expressed as follows:
|                    | Actual 0 (Negative) | Actual 1 (Positive) |
|--------------------|--------------------|---------------------|
| **Predicted 0**    | True Negative (TN) | False Negative (FN) |
| **Predicted 1**    | False Positive (FP)| True Positive (TP)  |

The diagonal contains the number of correctly classified samples.

Off the diagonal, the number of misclassified samples.

Where:

- **TP**: *True Positive* — positive samples correctly predicted as positive.  
- **FP**: *False Positive* — negative samples incorrectly predicted as positive.  
- **TN**: *True Negative* — negative samples correctly predicted as negative.  
- **FN**: *False Negative* — positive samples incorrectly predicted as negative.

⚠️ **Attention !**  
In **scikit-learn**, the confusion matrix layout is different from the usual convention.  
By default, `confusion_matrix(y_true, y_pred)` places:


So the confusion matrix looks like this:

|              | Predicted 0 | Predicted 1 |
|--------------|-------------|-------------|
| **Actual 0** | TN          | FP          |
| **Actual 1** | FN          | TP          |
 


#### Recall = TP / (TP + FN) 

**Recall** is appropriate when minimizing false negatives is important — that is, when the cost of missing positive instances is high.

For example, in a medical context, failing to detect a patient's disease (a false negative) can have serious consequences for the person.  
A model with high recall reduces the risk of missing a desired detection.

#### Precision = TP / (TP + FP)

**Precision** is appropriate when minimizing false positives is important.

For example, an email falsely identified as spam (a false positive) may cause important information to be lost.  
A model with high precision reduces the risk of false alarms.

#### The accuracy paradox

There is sometimes a strong imbalance between negative and positive cases in the dataset. For example, in machine failure detection, disease diagnosis, or fraud detection.

Take for example a fraud dataset where 1% of the samples correspond to positive cases (fraud present).

A simplistic and absurd prediction model that always predicts no fraud will have an accuracy of 99%. An excellent score but obviously useless. Hence the importance of having classification metrics adapted to each specific situation.

The scores obtained depend on the classification threshold mentioned earlier.

By default, a threshold of \( t = 0.5 \) is considered:  
if the predicted probability < 0.5, then the prediction is category 0, otherwise category 1. But other threshold values can be considered.

Let's compare the results obtained for two other thresholds, for example \( t = 0.3 \) and \( t = 0.7 \).

To get the predicted categories with these thresholds, we start from the predicted probabilities given by the model:

Finally, let's look at a more general classification metric that can be used in most cases: the ROC AUC.

The idea is to plot the recall or TPR (true positive rate) against the FPR (false positive rate) as a function of the classification thresholds. The TPR is defined as TP / (TP + FN) and the FPR is defined as FP / (FP + TN).

The closer the curve is to the top-left corner, the better the model. The diagonal corresponds to a purely random model.

It takes into account all classification thresholds between 0 and 1, and therefore does not depend on a specific threshold value.

## Clustering with K-means

So far, we have worked with supervised learning techniques (such as regression or classification) where the training dataset includes the target variable values.

Here, we will focus on an unsupervised learning technique: **clustering**.

The principle of clustering is to automatically group dataset samples that are similar. We partition the data into a finite number of coherent subsets, which can be called **“groups,” “classes,” or “categories.”**

The clustering algorithm has two main objectives:

- Minimize the difference between samples within the same group. We want dense and homogeneous groups.

- Maximize the difference between groups. We want groups that are well separated from each other.

To illustrate, here are three point clouds (or clusters):

- Cohesive, overlapping groups.

- Separated but very dispersed groups.

- Cohesive and well-separated groups.

Framing the problem like this immediately raises 3 questions.

**Question 1: How to define the similarity between samples?**

The similarity or resemblance between samples is directly expressed by a small distance between them. The most commonly used measure is the Euclidean distance.

**Question 2: How to find the optimal number of clusters?**

This is one of the model’s input parameters. You can decide to partition the samples into 2 groups, 3 groups, or more. This decision depends on the context, your knowledge of the data, and your objectives.

However, it is not always possible to know the optimal number of groups in advance. Therefore, we need to estimate this parameter before training the model.

To estimate the optimal number of groups, several methods exist, with poetic names such as the “Elbow Method” or the “Silhouette Method,” which we will see later.

**Question 3: What types of algorithms are used for clustering?**

We will just study k-means

### Understand k-means

The K-means algorithm is simple and effective. Here are its main steps:

Initialization phase:

- You arbitrarily choose a number of clusters, K.

- The model randomly selects K points as initial cluster centers, called centroids.

At each iteration:

 1. Assigning data points: Each data point is assigned to the cluster whose centroid is closest.

 2. Updating centroids: The coordinates of each cluster's centroid are recalculated based on the new points assigned to it.

This iterative process continues until the centroids no longer change significantly, or a predefined number of iterations is reached.

Once the process is complete, you obtain K clusters with similar data points grouped together.

K-means is a simple algorithm for grouping data, but it requires careful selection of the number of clusters and can be sensitive to the initial centroid placement.

The final result depends on:

- The number of clusters initially chosen;

- The initialization: in most implementations of K-means, the initial centroids are chosen automatically and randomly. This means you may get slightly different results each time the model is trained.

```python
k_means_labels = k_means.predict(X)

fig = plt.figure(figsize=(12, 6))
colors = ["#4EACC5", "#FF9C34", "#4E9A06","#111111"]
colors = ["#FF0000", "#0000FF", "#00FF00","#111111"]

# KMeans
ax = fig.add_subplot(1, 2, 1)

# for k, col in zip(range(n_clusters), colors):
for k in range(n_clusters):
    ax.plot(X[labels_true == k, 0], X[labels_true == k, 1], "w", markerfacecolor=colors[k], marker="o", markersize=6, alpha = 1)

ax.spines['bottom'].set_position('zero')
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_linestyle('dotted')
ax.spines['left'].set_linestyle('dotted')

# Customize the appearance of grid lines (dotted and alpha=0.5)
ax.xaxis.grid(True, linestyle='--', alpha=0.5)
ax.yaxis.grid(True, linestyle='--', alpha=0.5)
# Remove the top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_title("Original")

ax = fig.add_subplot(1, 2, 2)
for k in range(n_clusters):
    cluster_center = k_means.cluster_centers_[k]
    ax.plot(X[k_means_labels == k, 0], X[k_means_labels == k, 1], "w", markerfacecolor=colors[k], marker="o", markersize=6, alpha = 1)
for k in range(n_clusters):
    cluster_center = k_means.cluster_centers_[k]
    ax.plot(
    cluster_center[0],
    cluster_center[1],
    "o",
            markerfacecolor='#000',
            markeredgecolor=colors[k],
            markersize=9,
        )
ax.spines['bottom'].set_position('zero')
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_linestyle('dotted')
ax.spines['left'].set_linestyle('dotted')

# Customize the appearance of grid lines (dotted and alpha=0.5)
ax.xaxis.grid(True, linestyle='--', alpha=0.5)
ax.yaxis.grid(True, linestyle='--', alpha=0.5)
# Remove the top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_title("KMeans")
plt.tight_layout()
plt.show()
```

#### Method 3: Use the Silhouette Coefficient
The silhouette coefficient takes into account not only the density within clusters but also the separation between clusters.

It is calculated using the average intra-cluster distance (a) and the average nearest-cluster distance (b).

- a is the average distance between a sample and the other points in the same cluster (its own cluster centroid).

- b is the average distance between that sample and the points in the nearest neighboring cluster, i.e., the closest one it does not belong to.

- The silhouette coefficient for a sample is then defined as silhouette = (b - a) / max(a, b)

The silhouette score ranges from -1 (poor) to 1 (excellent).

In Python, you can compute it using silhouette_score from scikit-learn:

```python
from sklearn.metrics import silhouette_score
k_means_labels = k_means.predict(X)
print("silhouette_score: ", silhouette_score(X,k_means_labels ))
```

We obtain a silhouette coefficient of 0.60.

In this example, we are in an ideal case: we know the data, their respective groups, and the number of clusters. On top of that, the dataset is linearly separable—we can draw a straight line to separate each group of points.

In practice, it's rare that we know the number of clusters in advance. In such cases, we can also use the silhouette coefficient to determine the optimal number of clusters.

To illustrate this, let’s vary the number of clusters as a parameter of k-means, from 2 to 10, and observe how the silhouette coefficient evolves.

```python
centers = [[2, 2], [-2, -2], [2, -2]]
X, labels_true = make_blobs(n_samples=3000, centers=centers, cluster_std=0.9)

silhouette_scores = []
for n_clusters in range(2,8):
    print()
    print("n_clusters",n_clusters)
    k_means = KMeans(init="k-means++", n_clusters=n_clusters, random_state = 808 , n_init = 'auto'  )
    k_means.fit(X)
    print("score", k_means.score(X))

    k_means_labels = k_means.predict(X)
    print("silhouette_score: ", silhouette_score(X,k_means_labels ))
    silhouette_scores.append(silhouette_score(X,k_means_labels ))

fig = plt.figure(figsize=(9, 6))

ax = fig.add_subplot(1, 1, 1)
plt.plot(range(2,8), silhouette_scores)
ax.set_title('silhouette scores')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xlabel('Nombre de clusters')
ax.set_ylabel('Coefficient de silhouette')
plt.grid()
plt.show()
```

We clearly observe a peak in the silhouette coefficient when the number of clusters is 3.

### Apply k-means on iris
Iris is a famous dataset, there is 150 flower sample with 3 families of 50 flowers : 
- Iris-Setosa 

- Iris-Versicolor 

- Iris-Virginica

The 4 variables are the width and length, in cm, of the petals and sepals of the flowers. A picture is worth more than an explanation when it comes to the difference between a sepal and a petal



We can load the dataset like this

```python
from sklearn import datasets
iris = datasets.load_iris()
X = iris.data
y = iris.target
```

```python
print(X,y)
```

X contains the values of the predictors in the form of a NumPy array, and y is the list of classes: 0, 1, 2.

Let’s apply k-means to this dataset:

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=3, n_init="auto")
model.fit(X)
```

```python
# Exemple : une nouvelle fleur avec ses 4 caractéristiques
new_flower = [[5.1, 3.5, 1.4, 0.2]]  # liste dans une liste car sklearn attend un 2D array

# Prédire à quel cluster cette fleur appartient
cluster = model.predict(new_flower)

print(f"This flower is : {cluster[0]}")
```

Here are the classification scores:

```python
from sklearn.metrics import silhouette_score
print("score", model.score(X))
print("silhouette_score: ", silhouette_score(X,model.labels_ ))
```

Score: -78.85

Silhouette score: 0.55

Since we know the flowers belong to 3 families, we can also evaluate the clustering performance as a classification tool. Does the model correctly recover the original categories?

Note that, unlike supervised classification, this is not the primary goal of the clustering model. The clustering model only aims to group samples that are similar and to detect which arbitrary category they belong to.

Before doing this, we need to ensure that the cluster labels assigned by the model correspond correctly to the original categories. By using linear_sum_assignment function we can find the good labels.

```python
from sklearn.metrics import confusion_matrix, accuracy_score
from scipy.optimize import linear_sum_assignment
import numpy as np

# Confusion matrix between predicted and true labels
cm = confusion_matrix(y, model.labels_)

# Find the best label permutation
row_ind, col_ind = linear_sum_assignment(-cm)  # maximise matching

# Create a mapping from old labels to true labels
label_mapping = {old: new for old, new in zip(col_ind, row_ind)}
new_labels = np.array([label_mapping[label] for label in model.labels_])    

# Now evaluate
print("accuracy_score:", accuracy_score(y, new_labels))
print("confusion_matrix:\n", confusion_matrix(y, new_labels))
```

So, a fairly good performance with an accuracy of 88.7%.

The confusion matrix shows that category 1 is identified 100% correctly, and categories 2 and 3 are also mostly well identified.

The graphical analysis confirms this:

```python
import plotly.graph_objects as go
# plotting functions
def plot_3d(x, y, z, labels, title):
    return go.Scatter3d(
        x=x,
        y=y,
        z=z,
        mode='markers',
        marker=dict(
            size=5,
            color=labels,
            colorscale='Viridis',
            opacity=0.8
        ),
        name=title
    )

# Selection of 3 dimensions
x, y_, z = X[:, 0], X[:, 1], X[:, 2]

# plotting
fig = go.Figure()

fig.add_trace(plot_3d(x, y_, z, new_labels, "Clusters K-Means"))
fig.add_trace(plot_3d(x, y_, z, y, "real class"))

# formatting
fig.update_layout(
    title="3D visualization of clusters and real classes (Iris)",
    scene=dict(
        xaxis_title=iris.feature_names[0],
        yaxis_title=iris.feature_names[1],
        zaxis_title=iris.feature_names[2]
    ),
    legend=dict(itemsizing='constant'),
    margin=dict(l=0, r=0, b=0, t=40)
)

fig.show()
```
