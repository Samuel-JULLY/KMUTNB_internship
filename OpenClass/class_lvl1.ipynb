## Application of machine learning

Here is a dataset with 237 samples including age, sex, height, and weight of children aged 11.5 to 20 years.

Age is given in months, not years, ranging from 139 to 250 months (i.e., 11.5 to 20 years).
The sex variable is binary: 0 for boys and 1 for girls.
Height in cm ranges from 128.27 cm to 182.88 cm.
The target variable, weight in kg, ranges from 22.9 kg to 77.78 kg.
Below is a short example of how to train a predictive model on this data.

```python
import pandas as pd
import numpy as np
```

```python
# Load dataset in panda's dataframe
df = pd.read_csv('https://raw.githubusercontent.com/OpenClassrooms-Student-Center/8063076-Initiez-vous-au-Machine-Learning/master/data/age_vs_poids_vs_taille_vs_sexe.csv')
```

```python
# 5 first lines of the dataframe
df.head()
```

```python
# Statistics of the dataframe
df.describe()
```

```python
# predictive variables (also called features)
X = df[['sexe', 'age']]

# The target variable, weight
y = df.poids
```

```python
# Import and instantiate the model

from sklearn.linear_model import LinearRegression
reg = LinearRegression()
```

```python
# We train this model on the data using the fit method
reg.fit(X, y)
```

```python
# and we get the score
print(f"R^2 : {np.round(reg.score(X, y), 3)}")
```

```python
# and the a,b,c, coefficient of the linear regression
print(f" weight = {np.round(reg.coef_[0],  2)} * sex + {np.round(reg.coef_[1],  2)} * age  + noise")
```

```python
# features
X = df[['sexe', 'age', 'taille']]

# target
y = df.poids

# training of a new linear regression model
reg2 = LinearRegression()
reg2.fit(X,y)

# le score
print(f"R^2 : {np.round(reg2.score(X, y), 3)}")
# et les coefficients
print(f"weight = {np.round(reg.coef_[0],  2)} * sex + {np.round(reg.coef_[1],  2)} * age +  {np.round(reg2.coef_[2],  2)} * height + noise")
```

```python
# Weight prediction for a boy aged 150 months
reg.predict(np.array([[0, 150]]))
```

```python
# Weight prediction for a boy aged 150 months and 170 cm tall.
reg2.predict(np.array([[0, 150, 170]]))
```

```python
# Predictions on all the samples in the dataset.

y_pred = reg.predict(df[['sexe', 'age']])
y_pred2 = reg2.predict(df[['sexe', 'age', 'taille']])

from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error
print("1st Model: weight ~ sex + age")
print(f"\tmean_squared_error(y, y_pred): {mean_squared_error(y, y_pred)}")
print(f"\tmean_absolute_error(y, y_pred): {mean_absolute_error(y, y_pred)}")
print(f"\tmean_absolute_percentage_error(y, y_pred): {mean_absolute_percentage_error(y, y_pred)}")
print()



print("2nd Model: weight ~ sex + age + height")
print(f"\tmean_squared_error(y, y_pred): {mean_squared_error(y, y_pred2)}")
print(f"\tmean_absolute_error(y, y_pred): {mean_absolute_error(y, y_pred2)}")
print(f"\tmean_absolute_percentage_error(y, y_pred): {mean_absolute_percentage_error(y, y_pred2)}")
```

```python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
dataset_url = "https://raw.githubusercontent.com/OpenClassrooms-Student-Center/8063076-Initiez-vous-au-Machine-Learning/master/data/advertising.csv"

df = pd.read_csv(dataset_url)
```

```python
fig, ax = plt.subplots(1,3, figsize = (15,5))

plt.subplot(1,3,1)
sns.regplot(x = df[['tv']],y  =  df.ventes)
plt.ylabel('ventes')
plt.xlabel('TV')
plt.title('ventes = a * tv  + b')
plt.grid()
sns.despine()

plt.subplot(1,3,2)
sns.regplot(x = df[['radio']],y  =  df.ventes)
plt.ylabel('ventes')
plt.xlabel('radio')
plt.title('ventes = a * radio + b')
plt.grid()
sns.despine()

plt.subplot(1,3,3)
res = sns.regplot(x = df[['journaux']],y  =  df.ventes)
plt.ylabel('ventes')
plt.xlabel('journaux')
plt.title('ventes = a * journaux + b')
plt.grid()
sns.despine()

plt.tight_layout()
plt.show()
```

```python
df.corr()
```

```python
reg = LinearRegression()
```

```python
X = df[['tv','radio','journaux']]
y = df.ventes

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
# Entrainons le modele
reg.fit(X_train, y_train)
```

```python
y_pred_test = reg.predict(X_test)
```

```python
print(f"RMSE: {mean_squared_error(y_test, y_pred_test)}")
print(f"RMSE: {mean_absolute_percentage_error(y_test, y_pred_test)}")
```

```python
# Creer la variable quadratique tv^2
df['tv2'] = df.tv**2

variables = ['tv','radio','journaux','tv2']
```

```python
# Instancier le Scaler
scaler = MinMaxScaler()

# et l'appliquer 
data_array = scaler.fit_transform(df[variables])

# transformer le array en dataframe (c'est plus friendly)
df_scaled = pd.DataFrame(data_array,  columns = variables)

# verifier que les min et max des variables sont tous 0 et 1
df_scaled.describe().loc[['min','max']]
```

```python
# scindons les donnees en test et train
X = df_scaled[variables]
y = df.ventes

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
# entrainons le modele
reg = LinearRegression()
reg.fit(X_train, y_train)
```

```python
# Prediction sur le test set
y_pred_test = reg.predict(X_test)

# Scores
print(f"-- Regression ventes ~ radio + journaux + tv + tv^2")
print(f"\tRMSE: {mean_squared_error(y_test, y_pred_test)}")
print(f"\tMAPE: {mean_absolute_percentage_error(y_test, y_pred_test)}")
```

```python
# créons la variable tv * radio

df['tv_radio'] = df.tv * df.radio
```

```python
regressions = {
    'simple: y ~ tv + radio + journaux'  :     ['tv','radio','journaux'],
    'quadratique: y ~ tv + radio + journaux + tv2': ['tv','radio','journaux', 'tv2'],
    'terme croisée: y ~ tv + radio + journaux + tv*radio':['tv','radio','journaux', 'tv_radio']
}
```

```python
# la varibale cible est toujours la même
y = df.ventes


# boucle sur les régressions
for title, variables in regressions.items():
    # il faut limiter l'amplitude sur toutes les variables predictrices
    scaler = MinMaxScaler()
    data_array = scaler.fit_transform(df[variables])
    # df_scaled = pd.DataFrame(data_array, columns = variables)
    
    # X = df_scaled[variables]
    # Train test split
    X_train, X_test, y_train, y_test = train_test_split(data_array, y, test_size=0.20, random_state=42)

    # entrainer le modele
    reg.fit(X_train, y_train)
    # Prediction sur le test set
    y_pred_test = reg.predict(X_test)

    # Scores
    print(f"\n-- Regression {title}")
    print(f"\tRMSE: {mean_squared_error(y_test, y_pred_test)}")
    print(f"\tMAPE: {mean_absolute_percentage_error(y_test, y_pred_test)}")
```
